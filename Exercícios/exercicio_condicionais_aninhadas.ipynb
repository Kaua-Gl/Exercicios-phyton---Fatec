```python
#1-Sistema de emprestimo bancario condicionado
idade = int (input("Inisira sua idade: "))
renda_mensal = float (input("Insira sua renda mensal: "))

if idade >= 25 and idade <= 65 and renda_mensal >= 2500:
  valor_emprestimo = round (float (input("Insira o valor de emprestimo desejado: ")), 2)
  qtd_parcelas = int (input("Insira a quantidade de parcelas: "))
  valor_parcelas = round(valor_emprestimo / qtd_parcelas,2 )
  comprometimento = renda_mensal * 0.3

  if valor_parcelas < comprometimento:
    print(f"O emprestimo de R${valor_emprestimo} Com parcelas de {qtd_parcelas}x R${valor_parcelas} Foi  aprovado!")
  else:
    print("A solicitação não atende os requisitos minimos de comprometimento. (30%) Emprestimo negado.")
else:
  print("Não atende os requisitos de idade ou renda. Emprestimo negado.")
```
