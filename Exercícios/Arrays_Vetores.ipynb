```python
# Cadastro e Exibição Invertida (Índices e range regressivo)
numeros=[]
i=0
while i < 5:
    i += 1
    numero_cliente = float(input(f"Insira o {i}º numero"))
    numeros.append (numero_cliente)
i=0
for i in range(4, -1, -1):
    print(numeros[i])
    i+=1
```

```python
# Atualização de Preços na Lista
precos = [10.0, 25.0, 40.0, 15.0]
i=0
for i in range(0, 4):
    aumento = (precos[i] * (10/100)) + precos[i]
    precos[i] = aumento
    i+=1
print(precos)
```

```python
# Separador de Pares e Ímpares (Múltiplas Listas)
i=0
todos = []
impares = []
pares = []
for i in range (0, 6):
    numero = int(input(f"Insira o {i+1}º numero inteiro: "))
    todos.append (numero)

for i in  range (0, 6):
    if todos[i] % 2 == 0:
        pares.append (todos[i])
for i in  range (0, 6):
    if todos[i] % 2 != 0:
        impares.append (todos[i])

print("lista: ")
print(todos)
print("lista numeros pares: ")
print(pares)
print("lista numeros impares: ")
print(impares)
```

```python
#Buscador de Nomes (Pesquisa Linear e Flag Booleana)
alunos = ["Ana", "Bruno", "Carlos", "Diana", "Eduardo"]
pesquisa = str(input("pesquisa: "))
busca = False
i = 0
for i in range(0, 5):
    if alunos[i] == pesquisa:
        print(f"Aluno(a) {alunos[i]} foi encontrado na posição {i}")
        busca = True
if busca == False:
    print("Aluno(a) não encontrado")
```

```python
# Soma Posição por Posição entre Duas Listas
lista_a = [5, 6, 8, 1]
lista_b = [4, 23, 98, 43]
lista_c = []
i = 0
for i in range(0, 4):
    soma = lista_a[i] + lista_b[i]
    lista_c.append (soma)
print(f"a soma da lista a e lista b é: {lista_c}")
```
