```python
#validador de nota escolar
nota = float(input("Insira a sua nota: "))
while nota < 0 or nota > 10:
  print("Erro, o valor inserido é invalido")
  nota = float(input("Insira a sua nota: "))
print("Nota registrada no sistema!")
```

```python
#Caixa registradora com sentinela de saída
valor_total=0
valor=0
qtd_produtos=0
while valor != -1:
  valor = float(input("Insira o valor da compra "))
  if valor != -1:
    valor_total += valor
    qtd_produtos += 1
print(f"Quantidade de produtos: {qtd_produtos}, Total a pagar: R${valor_total}")
```

```python
#Contagem regressiva interativa
numero = int(input("Insira um numero para dar inicio a contagem: "))
while numero != 0:
  print(numero)
  numero += -1
  if numero == 0:
    print(0)
print("FIM DO TEMPO!")
```

```python
#Menu de Conversão de medidas
menu=0
valor=0
while menu != 3:
  print ("Insira 1 para converter (m) para (cm), 2 para converter (cm) para m, Insira 3 para Encerrar")
  menu = int(input())
  if menu == 1:
    valor = float(input("Insira o valor que deseja ser convertido"))
    conversao = valor * (10**2)
    print (f"{valor}(m) = {conversao}(cm)")
  elif  menu == 2:
    valor = float(input("Insira o valor que deseja ser convertido"))
    conversao = valor / (10**2)
    print (f"{valor}(cm) = {conversao}(m)")
```

```python
#Desafio da Tabuada com contagem de tentativas
resposta = 0
contagem = 0
while resposta != 1348:
  resposta = int(input("Quanto é 4 x 337: "))
  contagem += 1
  if resposta != 1348:
    print("Resposta errada!, Tente novamente")
print(f"Parabens!, Nº de tentativas: {contagem}")
```
