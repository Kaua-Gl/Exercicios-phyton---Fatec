```python
#exemplo de return
def soma(numero1, numero2):
    calculo = numero1 + numero2
    return calculo
resultado = soma(10,8)
print (resultado)

def media(soma):
    calculo = soma / 2
    return calculo
resultado_media = media(resultado)
print (resultado_media)
```

```python
#calcular área do retangulo
def calc_area_retangulo(valor_1, valor_2): 
    calculo = valor_1 * valor_2
    return calculo

base = float(input("Insira o valor da base do retangulo (m): "))
altura = float(input("Insira o valor da altura do retangulo (m): "))
resultado = calc_area_retangulo(base, altura)
print(f"A área do retango é de: {resultado} (m)")
```

```python
#Verificador de Par ou ímpar
def eh_par(valor):
    if valor % 2 != 0: 
        return "False"
    return "True"

numero = float(input("Digite um número: "))

if eh_par(numero) != "False":
   resultado = "par"
else:
    resultado = "ímpar"

print(f"O numero {numero} é {resultado}")
```

```python
#Situação final do aluno
def situacao_aluno(nota1,nota2):
    calc_media = (nota1+nota2) / 2
    if calc_media >= 7:
        return "Aprovado"
    return "Reprovado"

nota_1 = float(input("Insira a primeira nota: "))
nota_2 = float(input("Insira a primeira nota: "))

resultado = situacao_aluno(nota_1, nota_2)

print(f"O aluno foi {resultado}")
```
