```python
#Somatorio de numeros impares
i=0
soma=0
for i in range(1, 51, 2):
  soma += i
print(f"a soma dos numeros impares de 1 a 50 é: {soma}")
```

```python
#Fatorial acumulador
i=0
numero = int(input("insira um numero inteiro positivo: "))
numero_texto = numero
for i in range((numero-1), 1, -1):
  numero = numero * i
print(f"{numero_texto}! = {numero}")
```

```python
#contador de numeros negativos
i=0
contador=0
for i in range(0, 6):
  numero = int(input("insira um numero negativo ou positivo: "))
  if numero < 0:
    contador = contador + 1
  else:
    contador = contador
print (f"A quantidade de numeros negativos foi: {contador}")
```

```python
#Gerador de intervalo personalizado
i=0
inicio = int(input("insira um valor de inicio: "))
fim = int(input("insira um valor de fim: "))
for i in range((inicio-1), fim):
  i += 1
  print(i)
```
