```python
#Atividade 1

valor = float (input ("valor total Reais"))
cotacaodolar = float (input ("cotaçao Dólar"))
taxa_banco = 0.011
taxa = "1.1%"

conversora = valor / cotacaodolar
aplica_taxa = conversora * taxa_banco
valortotal = conversora + aplica_taxa
valortotal = round (valortotal,2)


print (f"Valor em real: R${valor}  Taxa Bancaria: {taxa}  Total convertido: ${valortotal}")
```

```python
# radar de velocidade
vel_carro = int (input ("insira a velocidade do carro"))
valor_multa = 100
if vel_carro > 80:
  taxa_multa = (vel_carro - 80) / 80
  calc_porcent = valor_multa * taxa_multa
  multa_total = valor_multa + calc_porcent

  print (f"R${multa_total}")
else:
  print ("velocidade dentro do limite")

```

```python
# média com frequencia
nota1 = float (input("insira a nota"))
nota2 = float (input("insira a nota"))
nota3 = float (input("insira a nota"))
frequencia = float (input("insira a frequencia"))

calc_media = (nota1 + nota2 + nota3) / 3
media_round  = round (calc_media,2)

if media_round >= 6 and frequencia >= 75:
  print (f"aprovado!: média: {media_round} frequencia: {frequencia}%")
else:
  print (f"reprovado!: média: {media_round} frequencia: {frequencia}%")
```

```python
# caixa eletronico seguro
usuario_sistema = "Kaua"
senha_sistema = "12345"
saldo = float(4300)

usuario = str (input("insira o usuario"))
senha = str (input("insira a senha"))

if usuario == usuario_sistema:

  if senha == senha_sistema:
    print ("acesso permitido")
    saque = float (input("insira o valor de saque"))

    if saldo < saque:
      print ("saldo insuficiente")
    else:
      print ("saque confirmado, verifique a saida do caixa")

  else:
    print ("usuario ou senha incorretos")

else:
  print ("usuario ou senha incorretos")
```

```python
# Calculadora de IMC
peso = float (input("inserir peso em Kg"))
altura = float (input("inserir altura em Metros"))

imc = peso / (altura ** 2)
imc_round = round (imc, 2)

if imc < 18.5:
  print (f"imc: {imc_round} abaixo do peso")

elif imc >= 18.5 and imc <= 24.9:
  print (f"imc: {imc_round} Peso normal (saudavel)")

elif imc >= 25.0 and imc <= 29.9:
  print (f"imc: {imc_round} Sobrepeso ")

elif imc >= 30.0 and imc <= 24.9:
  print (f"imc: {imc_round} Obesidade grau I")

elif imc >= 35.0 and imc <= 39.9:
  print (f"imc: {imc_round} Obesidade grau II")

else:
  print (f"imc: {imc_round} Obesidade grau III")


```
