```python
#sistema de desconto
valor = float(input("insira o valor do produto (R$) "))

if valor < 100:
  print(f"valor original: R${valor}, desconto: 0%, valor final R${valor}.")
elif valor <= 299.99:
  desconto = round(valor - (valor * 0.05), 2)
  print(f"valor original: R${valor}, desconto: 5%, valor final: R${desconto}.")
elif valor <= 499.99:
  desconto = round(valor - (valor * 0.1), 2)
  print(f"valor original: R${valor}, desconto: 10%, valor final: R${desconto}.")
else:
  desconto = round(valor - (valor * 0.15),2 )
  print(f"valor original: R${valor}, desconto: 15%, valor final: R${desconto}.")
```

```python
#calculadora completa
primeiro_num = float(input("insira o primeiro numero "))
operacao = str(input("insira a operação "))
segundo_num = float(input("insira o segundo numero "))

if segundo_num == 0 and operacao == "/":
  print("divisão por zero não permitida")

elif operacao == "+":
  soma = primeiro_num  + segundo_num
  print(soma)

elif operacao == "-":
  subtracao = primeiro_num  - segundo_num
  print(subtracao)

elif operacao == "x" or operacao == "*":
  multiplicacao = primeiro_num  * segundo_num
  print(multiplicacao)

elif operacao == "/" or operacao == "÷":
  divisao = primeiro_num  / segundo_num
  print(divisao)

else:
  print("operação invalida")
```

```python
#sistema de estacionamento
desconto = 0.2
valor_fixo = 8
valor_max = 40

qtd_horas = float (input("insira a quantidade de horas desejada "))
vip = str (input("cliente VIP? S/N "))

if vip == "S" or vip == "s":
  if qtd_horas <= 0:
    print("tempo invalido")
  else:
    if qtd_horas <= 1:
      valor_desc = (valor_fixo * desconto)
      valor_total = valor_fixo - valor_desc
      print(f"Valor antes do desconto: R${valor_fixo}")
      print(f"Desconto VIP: R${valor_desc}")
      print(f"Total: R${valor_total}")
    elif qtd_horas <= 3:
      adicional =  valor_fixo + ((qtd_horas - 1) *5)
      valor_desc = adicional * desconto
      valor_total = adicional - valor_desc
      print(f"Valor antes do desconto: R${adicional}")
      print(f"Desconto VIP: R${valor_desc}")
      print(f"Total: R${valor_total}")

    else:
      adicional =  valor_fixo + ((qtd_horas - 1) *4)
      if adicional >= valor_max:
        print(f"Valor antes do desconto: R${valor_max}")
        valor_desc = valor_max * desconto
        valor_total = valor_max - valor_desc
        print(f"Desconto VIP: R${valor_desc}")
        print(f"Total: R${valor_total}")
      else:
        adicional =  valor_fixo + ((qtd_horas - 1) *4)
        valor_desc = adicional * desconto
        valor_total = adicional - valor_desc
        print(f"Valor antes do desconto: R${adicional}")
        print(f"Desconto VIP: R${valor_desc}")
        print(f"Total: R${valor_total}")

elif vip == "N" or vip =="n":
  if qtd_horas <= 0 :
    print("tempo invalido")
  else:
    if qtd_horas <= 1:
      print(f"Total: R${valor_fixo}")


    elif qtd_horas <= 3:
      adicional =  valor_fixo + ((qtd_horas - 1) *5)
      print(f"Total: R${adicional}")

    else:
      adicional =  valor_fixo + ((qtd_horas - 1) *4)
      if adicional >= valor_max:
        print(f"Total: R${valor_max}")
      else:
        print(f"Total: R${adicional}")
else:
  print ("Resposta invalida")
```
