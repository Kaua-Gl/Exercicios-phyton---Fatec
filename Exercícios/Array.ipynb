```python
#exercicio 1 - [criar uma lista vazia e adicionar dados em um laço for e exibir em uma lista]

numeros = []
i=0
for i in range(0, 5):
    valor = int(input(f"insira o {i+1}º numero inteiro"))
    numeros.append(valor)
print(numeros[0])
print(numeros[1])
print(numeros[2])
print(numeros[3])
print(numeros[4])             
len (numeros)
```

```python
#exercicio 2 - cadastrar 4 nomes em uma lista e dar print no primeiro elemento e no ultimo elemento e exibir o total com len

nomes = []
i=0
for i in range(0,4):
    nome_cliente = str(input(f"Digite o {i+1}º nome: "))
    nomes.append(nome_cliente)
print(nomes[0])
print(nomes[len(nomes)-1])
print(len(nomes))
```

```python
#exercicio 3 - média com array
i=0
notas = []
for i in range(0, 4):
    notas_user = float(input(f"Insira a {i+1} nota: "))
    notas.append(notas_user)
i = 0
soma = 0
for i in range (0,4):
    soma += notas[i]
media = round(soma/len(notas),2)

print (f"A média final é {media}")
```
