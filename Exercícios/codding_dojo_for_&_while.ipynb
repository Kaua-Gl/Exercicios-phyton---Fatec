```python
#Validador de S e N
validador = str(input('Deseja continuar? (S/N)'))

while validador != 's' and validador != 'n':
  print('Opção inválida!')
  validador = str(input('Deseja continuar? (S/N)'))
if validador == 's':
  print('Prosseguindo para a proxima etapa')
else:
  print('Finalizando')
```

```python
#Meta do Cofrinho (meta R$100)
deposito=0
saldo=0
while saldo < 100:
    deposito = float(input("Insira o valor do deposito: "))
    saldo += deposito
print(f"Meta alcançada! o total guardado é de {saldo}")
```

```python
#adivinhe o numero secreto
secreto = 15
numero_informado = int(input("Adivinhe o numero secreto: "))
while numero_informado != secreto:
    if numero_informado < 15:
        print("Muito baixo, tente novamente!")
    elif numero_informado > 15:
        print("Muito alto, tente novamente!")
    numero_informado = int(input("Adivinhe o numero secreto: "))
print("Parabéns! Você acertou o numero secreto!!!")

    
```
